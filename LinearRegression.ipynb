```python
import numpy as np
```

BƯỚC 1: CHUẨN BỊ DỮ LIỆU:
3 đặc trưng - USD Index, Lạm phát, Giá dầu

```python
X = np.array([
[92.5, 2.1, 65.3], [93.2, 2.5, 67.2], [91.8, 2.3, 64.0],
[94.0, 2.8, 70.1], [95.2, 3.0, 72.5], [96.1, 3.2, 74.3],
[90.5, 1.8, 61.0], [92.0, 2.0, 63.2], [89.5, 1.5, 59.8],
[97.0, 3.5, 76.2], [95.8, 3.1, 73.8], [94.5, 2.9, 71.5],
[91.2, 2.2, 62.8], [90.0, 1.7, 60.5], [98.0, 3.7, 78.0],
[99.2, 4.0, 80.5], [88.5, 1.3, 58.0], [87.8, 1.1, 56.5],
[86.5, 1.0, 55.0], [100.0, 4.2, 82.0]
], dtype=float)

y = np.array([
1800, 1825, 1795, 1850, 1880, 1905, 1750, 1780, 1725, 1925,
1890, 1860, 1775, 1740, 1950, 1980, 1700, 1680, 1650, 2000
], dtype=float)
```

 BƯỚC 2: TIỀN XỬ LÝ - CHUẨN HÓA DỮ LIỆU (Z-SCORE)

```python
X_mean = X.mean(axis=0)
X_std = X.std(axis=0)
Xn = (X - X_mean) / X_std
```

 BƯỚC 3: HUẤN LUYỆN MÔ HÌNH

```python
class LinearRegressionGD:
  """
Cài đặt mô hình Hồi quy tuyến tính từ đầu bằng Gradient Descent.
"""
  def __init__(self, lr=0.05, epochs=1000):
    self.lr = lr
    self.epochs = epochs
    self.w = None
    self.b = 0.0

  def predict(self, X):
    return X @ self.w + self.b

  def mse(self, y_true, y_pred):
    return np.mean((y_true - y_pred)**2)

  def fit(self, X, y, verbose_every=100):
    m, n = X.shape # m: số mẫu, n: số đặc trưng
    self.w = np.zeros(n)
    self.b = 0.0
    for ep in range(1, self.epochs + 1):
      y_hat = self.predict(X)
      residual = y_hat - y

      dw = (2/m) * (X.T @ residual)
      db = (2/m) * np.sum(residual)

      self.w -= self.lr * dw
      self.b -= self.lr * db
      if verbose_every and ep % verbose_every == 0:
        print(f"Epoch {ep:4d} | MSE: {self.mse(y, y_hat):.2f}")
```

```python
print("--- Bắt đầu quá trình huấn luyện ---")
model = LinearRegressionGD(lr=0.05, epochs=1000)
model.fit(Xn, y, verbose_every=200)
print("--- Huấn luyện hoàn tất! ---")
```

BƯỚC 4: ĐÁNH GIÁ VÀ DỰ ĐOÁN

```python
print("\n--- KẾT QUẢ CUỐI CÙNG ---")
print(f"Trọng số w đã học: {model.w}")
print(f"Hệ số chặn b đã học: {model.b:.4f}")
# Dự đoán trên toàn bộ tập dữ liệu đã chuẩn hoá
y_pred = model.predict(Xn)
print("\n--- SO SÁNH 5 MẪU ĐẦU TIÊN ---")
print("Giá thực tế | Giá dự đoán")
print("----------------------------")
for i in range(5):
  print(f"{y[i]:<12.2f} | {y_pred[i]:.2f}")
```
